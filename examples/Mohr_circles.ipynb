```python
import numpy as np
import matplotlib.pyplot as plt
U = np.random.rand
π = np.arccos(-1) 
```

```python
r = 2
R = 8
n = 0 # hits
N = 10**4

for i in range(N):
    x = 10*U() - 5
    y = 5*U()
    if (x + 4)**2 + y**2 > 1**2 and (x - 1)**2 + y**2 > 4**2 and x**2 + y**2 < 5**2: 
        n += 1
    
V = 50
p = n/N
p_true = 0.5*(π*25 - π - 16*π)/V # = 4π/50 ~= 0.251

I       = V*p
δI      = V*((1-p)*p/N)**0.5
I_true  = V*p_true # = 4π ~= 12.566 
δI_true = V*((1-p_true)*p_true/N)**0.5 # ~= 0.217

print('p = %2.3f vs %2.3f (exact)'%(p, p_true))
print('I = %2.3f +/- %2.3f vs %2.3f +/- %2.3f (expected)'%(I, δI, I_true, δI_true))
```
